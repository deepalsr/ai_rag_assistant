```python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()  # reads your .env file and loads GROQ_API_KEY into environment

client = Groq(api_key=os.environ["GROQ_API_KEY"])
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[{"role": "user", "content": "Say 'connection successful' and nothing else."}],
)

print(response.choices[0].message.content)
```

```python
question = "how many vacation days do I get?"

retrieved_chunks = [
    "SECTION 2: LEAVE POLICY\nFull-time employees accrue 18 days of paid time off (PTO) per year, accrued monthly at a rate of 1.5 days per month. PTO requests should be submitted through the HR portal at least 2 weeks in advance for planned leave.",
    "PTO up to 5 days may be carried over to the following year; anything beyond that is forfeited."
]

context = "\n\n---\n\n".join(retrieved_chunks)

prompt = f"""Answer the question using ONLY the context below. If the context doesn't contain the answer, say "I don't have enough information to answer that."

Context:
{context}

Question: {question}

Answer:"""

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[{"role": "user", "content": prompt}],
    temperature=0.1,
)

print(response.choices[0].message.content)
```
